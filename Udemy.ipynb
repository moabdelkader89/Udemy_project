lectures col

```python
#this fun takes parameters x-axis,y-axis,hue
def drew_fun(x,y,z):
    plt.figure(figsize=(7,5))
    sns.barplot(x=x,y=y,data=df,hue=z,palette=colors,errorbar=None)
    plt.title("    " + x +"  "+ "per to  "+y)    
    plt.figure(figsize=(5,5))
    df.groupby(x)[y].count().plot(kind='pie',autopct="%1.1f%%")
    plt.title("    " + x +"  "+ "per to  "+y)
    plt.show()
```

```python
drew_fun('level','num_subscribers','level')
```

Price And Income

Average of prices per courses

```python
#Clean the price col from some value its(true and free) by =0
#we can use this also > df['price'].str.replace('true|free','0',regex=true)
df['price']=df['price'].replace('Free','0')
df['price']=df['price'].replace('TRUE','0')
df['price']=df['price'].replace('true','0')

#then replace the datatype of this col from obj to int64
df['price']=df['price'].astype('Int64')

#average of courses price
ave_price=df.groupby('subject')['price'].mean()
ave_price
```

```python
# Average off All Courses
ave_all=df['price'].mean()
ave_all
```

```python
ave_price=df.groupby('subject')['price'].mean().plot(kind='bar')
plt.title('Average of prices per courses')
plt.xticks(rotation=45)
plt.show()
```
